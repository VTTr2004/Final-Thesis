```python
import cv2
import os
```

```python
def extract_frames(video_path : str, output_folder : str):
    os.makedirs(output_folder, exist_ok=True)
    
    cap = cv2.VideoCapture(video_path)
    if not cap.isOpened():
        print(f"Can't open the video: {video_path}")
        return
    
    fps = cap.get(cv2.CAP_PROP_FPS)  # FPS gốc của video
    frame_interval = int(fps * 3) 
    
    frame_idx = 0
    saved_idx = 0
    
    while True:
        ret, frame = cap.read()
        if not ret:
            break
        
        if frame_idx % frame_interval == 0:
            frame_path = os.path.join(output_folder, f"frame_{saved_idx:06d}.jpg")
            cv2.imwrite(frame_path, frame)
            saved_idx += 1

        frame_idx += 1
        
    cap.release()
    print(f"Done - Total frames: {frame_idx}")
    
def scan_folder(root, dest):
    videos_path = [os.path.join(root, path).replace("\\", "/") for path in os.listdir(root)]
    
    for i, path in enumerate(videos_path, 1):
        output = os.path.join(dest, str(i))
        os.makedirs(output, exist_ok=True)
        
        extract_frames(video_path = path, output_folder = output)
        
```

```python
scan_folder("data/VideoSource", "data/BEV")
```

```python
import os

def rename_all(folder_path, prefix="img"):
    files = sorted(os.listdir(folder_path))
    
    for idx, filename in enumerate(files, start=1):
        ext = os.path.splitext(filename)[1]     # lấy phần .jpg/.png...
        new_name = f"{prefix}_{idx:06d}{ext}"
        
        old_path = os.path.join(folder_path, filename)
        new_path = os.path.join(folder_path, new_name)
        
        os.rename(old_path, new_path)

    print(f"Done rename {len(files)} files.")

# Ví dụ:
for i in [1, 2, 3, 4, 5, 7, 8, 9]:
    rename_all(f"data/BEV_300/{str(i)}", prefix=f"cam{str(i)}")
```
